# `slice_max` / `slice_min` — Group-Aware Top & Bottom Slicing

`slice_max` and `slice_min` select the top or bottom N rows ordered by a column, with complete group awareness via `by=`.

| Method | Key Params | Purpose |
|---|---|---|
| `.pt.slice_max(col, n=1, by=None, with_ties=False)` | `col`, `n`, `by`, `with_ties` | Select top N rows per group |
| `.pt.slice_min(col, n=1, by=None, with_ties=False)` | `col`, `n`, `by`, `with_ties` | Select bottom N rows per group |

---

```python
import sys, os
_src = os.path.abspath(os.path.join(os.getcwd(), '..', '..', 'src'))
if not os.path.exists(_src): _src = os.path.abspath(os.path.join(os.getcwd(), '..', 'src'))
if _src not in sys.path: sys.path.insert(0, _src)

import pytae as pt

penguins = pt.sample_data['penguins']
tips = pt.sample_data['tips']
```

## 1. `slice_max()` — Top N Records

Retrieve the rows with the highest values in `col`.

```python
# Top 3 heaviest penguins overall
(
    penguins
    .pt.slice_max("body_mass_g", n=3)
    .pt.select("species", "island", "body_mass_g")
)
```

## 2. `slice_min()` — Bottom N Records

Retrieve the rows with the lowest values in `col`.

```python
# 3 lightest penguins overall
(
    penguins
    .pt.slice_min("body_mass_g", n=3)
    .pt.select("species", "island", "body_mass_g")
)
```

## 3. Group-Aware Slicing (`by=`)

When `by=` is specified, `slice_max` and `slice_min` extract top or bottom N rows **per group**.

```python
# Top heaviest penguin for each species
(
    penguins
    .pt.slice_max("body_mass_g", n=1, by="species")
    .pt.select("species", "island", "body_mass_g")
)
```

```python
# Lightest 2 penguins per island
(
    penguins
    .pt.slice_min("body_mass_g", n=2, by="island")
    .pt.select("island", "species", "body_mass_g")
)
```

## 4. Functional Style (`pt.slice_max`, `pt.slice_min`)

Both functions are available as standalone functional verbs `pt.slice_max(df, ...)` and `pt.slice_min(df, ...)`:

```python
# Functional slice_max by group
pt.slice_max(tips, "tip", n=1, by="day")[["day", "total_bill", "tip"]]
```
